### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()

```

#1. Análise do faturamento


```python
#Faturamento da loja1
listaSoma = []
soma_precos_loja1 = loja['Preço'].sum()
listaSoma.append(soma_precos_loja1)
print(f"A soma de todos os preços da Loja 1 é: R$ {soma_precos_loja1:.2f}")
#Faturamento da loja2
soma_precos_loja2 = loja2['Preço'].sum()
listaSoma.append(soma_precos_loja2)
print(f"A soma de todos os preços da Loja 2 é: R$ {soma_precos_loja2:.2f}")
#Faturamento da loja3
soma_precos_loja3 = loja3['Preço'].sum()
listaSoma.append(soma_precos_loja3)
print(f"A soma de todos os preços da Loja 3 é: R$ {soma_precos_loja3:.2f}")
#Faturamento da loja4
soma_precos_loja4 = loja4['Preço'].sum()
listaSoma.append(soma_precos_loja4)
print(f"A soma de todos os preços da Loja 4 é: R$ {soma_precos_loja4:.2f}")
print([float(item) for item in listaSoma])

# Nomes das lojas para o eixo x
nomes_lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

# Criar a figura e os eixos
plt.figure(figsize=(8, 6))
plt.bar(nomes_lojas, listaSoma, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])

# Adicionar rótulos e título
plt.xlabel('Lojas')
plt.ylabel('Faturamento Total *Mi (R$)')
plt.title('Comparação de Faturamento entre Lojas')
```

# 2. Vendas por Categoria


```python
#Vendas por categoria - loja1
lista_categoria = loja['Categoria do Produto']

def contagem_Produtos(lista):
  dict_contagem = {}
  for item in lista:
    if item in dict_contagem:
      dict_contagem[item] += 1
    else:
      dict_contagem[item] = 1
  return dict_contagem
resultado_contagem1 = contagem_Produtos(lista_categoria)
rotulo1 = list(resultado_contagem1.keys())
tamanho1 = list(resultado_contagem1.values())

print(f'Vendas da loja 1: {resultado_contagem1}: ')

#Vendas por categoria - loja2
lista_categoria = loja2['Categoria do Produto']

def contagem_Produtos(lista):
  dict_contagem = {}
  for item in lista:
    if item in dict_contagem:
      dict_contagem[item] += 1
    else:
      dict_contagem[item] = 1
  return dict_contagem
resultado_contagem2 = contagem_Produtos(lista_categoria)
rotulo2 = list(resultado_contagem2.keys())
tamanho2 = list(resultado_contagem2.values())
print(f'Vendas da loja 2: {resultado_contagem2}: ')
#Vendas por categoria - loja3
lista_categoria = loja3['Categoria do Produto']

def contagem_Produtos(lista):
  dict_contagem = {}
  for item in lista:
    if item in dict_contagem:
      dict_contagem[item] += 1
    else:
      dict_contagem[item] = 1
  return dict_contagem
resultado_contagem3 = contagem_Produtos(lista_categoria)
rotulo3 = list(resultado_contagem3.keys())
tamanho3 = list(resultado_contagem3.values())
print(f'Vendas da loja 3: {resultado_contagem3}')
#Vendas por categoria - loja4
lista_categoria = loja4['Categoria do Produto']

def contagem_Produtos(lista):
  dict_contagem = {}
  for item in lista:
    if item in dict_contagem:
      dict_contagem[item] += 1
    else:
      dict_contagem[item] = 1
  return dict_contagem
resultado_contagem4 = contagem_Produtos(lista_categoria)
rotulo4 = list(resultado_contagem4.keys())
tamanho4 = list(resultado_contagem4.values())
print(f'Vendas da loja 4: {resultado_contagem4}')

def criar_grafico_pizza(tamanhos, rotulos, titulo='Distribuição', cores=None):

    plt.figure(figsize=(3, 3))
    plt.pie(tamanhos,
          labels=rotulos,
          colors=cores,
          autopct='%1.1f%%',
          shadow=True,
          startangle=90,
          )
    plt.axis('equal')
    plt.title('Relação por item vendido')
    plt.show()




```
